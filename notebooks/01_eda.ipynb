# PolyAletheia - Exploratory Data Analysis

Exploring the NeurIPS Open Polymer Prediction 2025 dataset.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from smiles_func import get_atom_count
from data_loader import load_data, clean_data

# load and clean
df = load_data()
df = clean_data(df)
print(f"Dataset size: {len(df)}")
```

## 1. Property Distributions
How much data do we have for each property?

```python
props = ["Tg", "FFV", "Tc", "Density", "Rg"]

plt.figure(figsize=(15, 10))
for i, p in enumerate(props, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df[p].dropna(), kde=True)
    plt.title(f"{p} Distribution\n(n={df[p].notna().sum()})")
plt.tight_layout()
plt.show()
```

## 2. SMILES Length Analysis
How long are the polymer strings? This determines our tokenizer context window.

```python
# add length column
df["smiles_len"] = df["SMILES"].apply(len)
df["atom_count"] = df["SMILES"].apply(get_atom_count)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(df["smiles_len"])
plt.title("SMILES String Characters")

plt.subplot(1, 2, 2)
sns.histplot(df["atom_count"])
plt.title("Atom Count per Polymer")
plt.show()

print(f"Max SMILES length: {df['smiles_len'].max()}")
print(f"Max Atom count: {df['atom_count'].max()}")
```

## 3. Correlation Matrix
Do properties correlate with each other? (e.g. higher density -> lower FFV?)

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df[props].corr(), annot=True, cmap="coolwarm", center=0)
plt.title("Property Correlation Matrix")
plt.show()
```
